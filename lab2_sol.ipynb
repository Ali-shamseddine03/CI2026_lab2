```python
def fitness(solution: set[frozenset]) -> float:
    weight = sum(XCOST[i] for i in solution)
    covered = frozenset.union(*solution)
    assert covered == frozenset(OBJECTS)
    return - weight
```

```python
def valid(solution, objects) -> bool:
    if not solution:
        return False
    covered = frozenset.union(*solution)
    return covered == frozenset(objects)

def tweak(solution, sets) -> set[frozenset]:
        new_solution = set(solution)

        s = sample(sets, k=1)[0]

        if s in new_solution:
            new_solution.remove(s)
        else:
            new_solution.add(s)
        return new_solution
```

```python
results = []

for N in range(1, 101):

    OBJECTS = {s for s in range(N)}

    SETS = tuple(frozenset(sample(range(N), k=s+1)) for s in range(N))

    COSTS = tuple(10*random() + (s+random())**2 for s in range(N))

    XCOST = {SETS[i]: COSTS[i] for i in range(N)}

    def fitness(solution):
        return -sum(XCOST[s] for s in solution)

    solution = set(SETS)

    for _ in range(1000):
        new_solution = tweak(solution, SETS)

        if fitness(new_solution) > fitness(solution):
            if valid(new_solution, OBJECTS):
                solution = new_solution

    results.append((N, -fitness(solution), len(solution)))

    print(
        f"S{N}: cost = {-fitness(solution):.2f}, "
        f"selected sets = {len(solution)}"
    )
```
